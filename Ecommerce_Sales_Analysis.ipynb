# E-Commerce Sales Analysis

## Problem Statement
E-commerce businesses generate large amounts of sales data.
Analyzing this data helps understand sales performance,
customer purchasing patterns, product performance and regional trends.

## Objective
To analyze e-commerce sales data and identify useful business insights
using Python and data visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Order_ID': [
        1001, 1002, 1003, 1004, 1005,
        1006, 1007, 1008, 1009, 1010,
        1011, 1012, 1013, 1014, 1015
    ],

    'Date': [
        '2026-01-05', '2026-01-12', '2026-01-20',
        '2026-02-03', '2026-02-15', '2026-02-24',
        '2026-03-02', '2026-03-10', '2026-03-18',
        '2026-04-05', '2026-04-12', '2026-04-20',
        '2026-05-03', '2026-05-15', '2026-05-25'
    ],

    'Product': [
        'Laptop', 'Headphones', 'T-Shirt',
        'Smartphone', 'Shoes', 'Keyboard',
        'Laptop', 'Jeans', 'Smartwatch',
        'Headphones', 'Shoes', 'Smartphone',
        'T-Shirt', 'Keyboard', 'Smartwatch'
    ],

    'Category': [
        'Electronics', 'Electronics', 'Fashion',
        'Electronics', 'Fashion', 'Electronics',
        'Electronics', 'Fashion', 'Electronics',
        'Electronics', 'Fashion', 'Electronics',
        'Fashion', 'Electronics', 'Electronics'
    ],

    'Quantity': [
        1, 2, 3, 1, 2,
        2, 1, 2, 1, 3,
        1, 1, 4, 2, 2
    ],

    'Price': [
        55000, 2500, 800, 30000, 3000,
        1500, 55000, 2000, 8000, 2500,
        3000, 30000, 800, 1500, 8000
    ],

    'Region': [
        'South', 'North', 'South',
        'West', 'East', 'South',
        'North', 'West', 'South',
        'East', 'South', 'North',
        'West', 'East', 'South'
    ]
}

df = pd.DataFrame(data)

df
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df['Sales'] = df['Quantity'] * df['Price']

df.head()
```

```python
df['Month'] = df['Date'].dt.month_name()

df.head()
```

```python
df.shape
```

```python
df.head()
```

```python
total_revenue = df['Sales'].sum()

print("Total Revenue: ₹", total_revenue)
```

```python
total_orders = df['Order_ID'].nunique()

print("Total Orders:", total_orders)
```

```python
total_units = df['Quantity'].sum()

print("Total Units Sold:", total_units)
```

```python
product_sales = df.groupby('Product')['Sales'].sum()

print(product_sales)
```

```python
top_product = product_sales.idxmax()
top_product_sales = product_sales.max()

print("Highest Revenue Product:", top_product)
print("Revenue: ₹", top_product_sales)
```

```python
product_quantity = df.groupby('Product')['Quantity'].sum()

best_selling_product = product_quantity.idxmax()
best_selling_quantity = product_quantity.max()

print("Best-Selling Product:", best_selling_product)
print("Units Sold:", best_selling_quantity)
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales)
```

```python
top_category = category_sales.idxmax()
top_category_sales = category_sales.max()

print("Top Category:", top_category)
print("Category Revenue: ₹", top_category_sales)
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales)
```

```python
top_region = region_sales.idxmax()
top_region_sales = region_sales.max()

print("Top Sales Region:", top_region)
print("Region Revenue: ₹", top_region_sales)
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

```python
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print("Best Sales Month:", best_month)
print("Monthly Revenue: ₹", best_month_sales)
```

```python
average_order_value = total_revenue / total_orders

print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
print("===== E-COMMERCE SALES SUMMARY =====")

print("Total Revenue: ₹", total_revenue)
print("Total Orders:", total_orders)
print("Total Units Sold:", total_units)

print("Highest Revenue Product:", top_product)
print("Best-Selling Product:", best_selling_product)

print("Top Category:", top_category)
print("Top Region:", top_region)
print("Best Sales Month:", best_month)

print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
product_sales.plot(kind='bar')

plt.title('Revenue by Product')
plt.xlabel('Product')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_sales.plot(kind='bar')

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales.plot(kind='bar')

plt.title('Revenue by Region')
plt.xlabel('Region')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
monthly_sales_ordered = (
    df.groupby(df['Date'].dt.to_period('M'))['Sales']
    .sum()
)

monthly_sales_ordered.index = monthly_sales_ordered.index.astype(str)

monthly_sales_ordered.plot(
    kind='line',
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_quantity.plot(kind='bar')

plt.title('Units Sold by Product')
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Key Business Insights

1. The business generated strong revenue mainly from Electronics.
2. Laptop was one of the major revenue-generating products.
3. Product sales varied significantly based on product price and quantity.
4. Regional analysis showed which region contributed the highest revenue.
5. Monthly analysis helped identify the best-performing sales month.
6. Best-selling products by quantity were different from products generating the highest revenue.
